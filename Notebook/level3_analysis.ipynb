```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
df.head()
```

```python
# Check missing values
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Calculate missing value percentage
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing value percentage:")
print(missing_percentage)
```

```python
# Display only columns that contain missing values
missing_summary = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Percentage': missing_percentage
})

missing_summary = missing_summary[missing_summary['Missing Values'] > 0]

print(missing_summary)
```

```python
print(df.columns.tolist())
```

```python
# Convert Purchase Date into datetime format
df['Purchase Date'] = pd.to_datetime(df['Purchase Date'], errors='coerce')

print("Purchase Date data type:")
print(df['Purchase Date'].dtype)
```

```python
# Create date-based features
df['Year'] = df['Purchase Date'].dt.year
df['Month'] = df['Purchase Date'].dt.month
df['Quarter'] = df['Purchase Date'].dt.quarter
df['Day'] = df['Purchase Date'].dt.day
df['Day_of_Week'] = df['Purchase Date'].dt.dayofweek
```

```python
df[['Purchase Date', 'Year', 'Month', 'Quarter', 'Day', 'Day_of_Week']].head()
```

```python
print("Updated Dataset Shape:", df.shape)
df.head()
```

```python
print("Numeric columns summary:")
print(df[['Price', 'Quantity', 'Total Sales', 'Customer Age']].describe())
```

```python
df['Average_Order_Value'] = df['Total Sales'] / df['Quantity']
```

```python
df[['Total Sales', 'Quantity', 'Average_Order_Value']].head()
```

```python
customer_frequency = df.groupby('Customer ID').size().reset_index(name='Purchase_Frequency')

df = df.merge(customer_frequency, on='Customer ID', how='left')
```

```python
df[['Customer ID', 'Purchase_Frequency']].head()
```

```python
customer_spending = df.groupby('Customer ID')['Total Sales'].sum().reset_index(name='Customer_Total_Spending')

df = df.merge(customer_spending, on='Customer ID', how='left')
```

```python
df[['Customer ID', 'Customer_Total_Spending']].head()
```

```python
print("New features created:")

print(df[
    [
        'Average_Order_Value',
        'Purchase_Frequency',
        'Customer_Total_Spending'
    ]
].head())
```

```python
print("Dataset Shape:", df.shape)
print("Number of Features:", len(df.columns))
```

```python
unique_customers = df['Customer ID'].nunique()

print("Total Unique Customers:", unique_customers)
```

```python
top_customers = (
    df.groupby('Customer ID')['Total Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("Top 10 Customers by Total Spending:")
print(top_customers)
```

```python
frequent_customers = (
    df.groupby('Customer ID')
    .size()
    .sort_values(ascending=False)
    .head(10)
)

print("Top 10 Customers by Purchase Frequency:")
print(frequent_customers)
```

```python
gender_analysis = df.groupby('Customer Gender')['Total Sales'].sum()

print("Sales by Customer Gender:")
print(gender_analysis)
```

```python
df['Age_Group'] = pd.cut(
    df['Customer Age'],
    bins=[0, 18, 30, 45, 60, 100],
    labels=['Under 18', '18-30', '31-45', '46-60', '60+']
)
```

```python
age_group_sales = (
    df.groupby('Age_Group', observed=False)['Total Sales']
    .sum()
)

print("Sales by Age Group:")
print(age_group_sales)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    gender_analysis.index,
    gender_analysis.values
)

plt.title("Sales by Customer Gender")
plt.xlabel("Customer Gender")
plt.ylabel("Total Sales")

plt.grid(axis='y')
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    age_group_sales.index.astype(str),
    age_group_sales.values
)

plt.title("Sales by Customer Age Group")
plt.xlabel("Age Group")
plt.ylabel("Total Sales")

plt.grid(axis='y')
plt.tight_layout()

plt.show()
```

```python
print("========== BUSINESS KPIs ==========")

total_sales = df['Total Sales'].sum()
total_quantity = df['Quantity'].sum()
average_order_value = df['Total Sales'].mean()
total_orders = df['Order ID'].nunique()
total_customers = df['Customer ID'].nunique()
total_products = df['Product Name'].nunique()

print("Total Sales:", total_sales)
print("Total Quantity Sold:", total_quantity)
print("Average Order Value:", average_order_value)
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Total Products:", total_products)
```

```python
monthly_sales = (
    df.groupby(df['Purchase Date'].dt.to_period('M'))['Total Sales']
    .sum()
)

print("Monthly Sales:")
print(monthly_sales.head(12))
```

```python
yearly_sales = (
    df.groupby('Year')['Total Sales']
    .sum()
)

print("Yearly Sales:")
print(yearly_sales)
```

```python
category_sales = (
    df.groupby('Category')['Total Sales']
    .sum()
    .sort_values(ascending=False)
)

print("Sales by Category:")
print(category_sales)
```

```python
top_products = (
    df.groupby('Product Name')['Total Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("Top 10 Products:")
print(top_products)
```

```python
best_category = category_sales.idxmax()
best_category_sales = category_sales.max()

print("Best Performing Category:", best_category)
print("Category Sales:", best_category_sales)
```

```python
best_month = monthly_sales.idxmax()

print("Best Sales Month:", best_month)
print("Highest Monthly Sales:", monthly_sales.max())
```

```python
best_year = yearly_sales.idxmax()

print("Best Sales Year:", best_year)
print("Highest Yearly Sales:", yearly_sales.max())
```
